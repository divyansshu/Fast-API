```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score
import joblib
```

```python
iris = load_iris()
```

```python
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df.head()
```

```python
df.columns
```

```python


# Example of renaming specific columns
df_renamed = df.rename(columns={
    'sepal length (cm)': 'SepalLengthCm',
    'sepal width (cm)': 'SepalWidthCm',
    'petal length (cm)': 'PetalLengthCm',
    'petal width (cm)': 'PetalWidthCm'
})

print("\nRenamed column names:")
print(df_renamed.columns)
```

```python
df_renamed
```

```python
print("Target column:")
print(iris.target_names)
```

```python
df_renamed['Species'] = iris.target
df_renamed
```

```python
df_renamed['Species'].unique()
```

```python
y = df_renamed['Species']
```

```python
X = df_renamed.drop(columns='Species')
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify = y, random_state=42)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
model = LogisticRegression()
model.fit(X_train.values, y_train.values)
```

```python
y_train_pred = model.predict(X_train.values)
y_test_pred = model.predict(X_test.values)
```

```python
print(f'Train Accuracy: {accuracy_score(y_train, y_train_pred)}')
print(f'Test Accuracy: {accuracy_score(y_test, y_test_pred)}')
```

```python
print('Classification Report')
print(classification_report(y_test, y_test_pred))
```
